```python
import pandas as pd
```

```python
df=pd.read_csv("netflix_titles.csv")

df.head()
```

```python

#check the columns and missing values

df.info()

#summary 
df.describe()

# checking number of rows and columns 

print(df.shape)
```

```python
#checking for duplicates

df.duplicated().sum()
```

```python
#filling the missing values with 'unknown'

df['director']=df['director'].fillna('unknown')
df['cast']=df['cast'].fillna('unknown')
df['country']=df['country'].fillna('unknown')
```

```python
#DROPPING MISSING DATES,RATINGS AND DURATION

df=df.dropna(subset=['date_added','rating','duration'])
```

```python
#verifying 

df.isnull().sum()
```

```python
#fixing mix type of values 

df['date_added']=pd.to_datetime(df['date_added'].str.strip())

df['month_added']=df['date_added'].dt.month_name()

df['year_added']=df['date_added'].dt.year 
```

```python
#fixing duration in minutes or in seasons

df[['duration_num','duration_unit']]=df['duration'].str.split(' ',expand=True)

df['duration_num'] = pd.to_numeric(df['duration_num'])
```

```python
df[['title', 'date_added', 'year_added', 'duration_num', 'duration_unit']].head()
```

```python
#exporting cleaned dataframe into a csv file 

df.to_csv("netflix_titles_cleaned.csv", index=False)
```
